```python
"""
============================================================
                    GEMINI AI CHATBOT
============================================================

Features:
    1. Normal Gemini text chat
    2. Conversation memory
    3. Image analysis
    4. PDF analysis
    5. Text/CSV/JSON/Markdown file analysis
    6. Multiple-file analysis
    7. AI image generation
    8. PDF report generation
    9. Save Gemini answers as TXT
   10. No Streamlit
   11. No PowerShell API-key configuration

Install:

    python -m pip install google-genai reportlab pillow

IMPORTANT:
    Put your Gemini API key below.

============================================================
"""


# ============================================================
# IMPORTS
# ============================================================

import os
from pathlib import Path
from datetime import datetime

from google import genai

from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.enums import TA_CENTER
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    Image as PDFImage,
    PageBreak,
)
from reportlab.lib.units import inch

from PIL import Image


# ============================================================
# 🔑 GEMINI API KEY
# ============================================================

GEMINI_API_KEY = "AQ.Ab8RN6LzlOW-ycmrCGTOxXC3HTjpRJ02VDxFY1dRvzZg9S3-7w"


# ============================================================
# MODELS
# ============================================================

TEXT_MODEL = "gemini-3.6-flash"

IMAGE_MODEL = "gemini-3.1-flash-image"


# ============================================================
# OUTPUT FOLDER
# ============================================================

OUTPUT_FOLDER = Path("gemini_output")

OUTPUT_FOLDER.mkdir(
    exist_ok=True
)


# ============================================================
# CHECK API KEY
# ============================================================

if (
    not GEMINI_API_KEY
    or GEMINI_API_KEY
    == "PASTE_YOUR_GEMINI_API_KEY_HERE"
):

    print()
    print("=" * 70)
    print("ERROR")
    print("=" * 70)
    print()
    print(
        "Please put your Gemini API key at the top of chatbot.py."
    )
    print()
    print(
        'Example: GEMINI_API_KEY = "YOUR_API_KEY"'
    )
    print()

    raise SystemExit


# ============================================================
# CREATE GEMINI CLIENT
# ============================================================

client = genai.Client(
    api_key=GEMINI_API_KEY
)


# ============================================================
# CONVERSATION MEMORY
# ============================================================

conversation_history = []


# ============================================================
# NORMAL TEXT CHAT
# ============================================================

def chat_with_gemini(user_message):

    global conversation_history

    conversation_history.append(
        {
            "role": "user",
            "content": user_message
        }
    )

    prompt = ""

    for message in conversation_history:

        if message["role"] == "user":

            prompt += (
                "User: "
                + message["content"]
                + "\n\n"
            )

        else:

            prompt += (
                "Assistant: "
                + message["content"]
                + "\n\n"
            )

    response = client.models.generate_content(
        model=TEXT_MODEL,
        contents=prompt
    )

    answer = response.text

    conversation_history.append(
        {
            "role": "assistant",
            "content": answer
        }
    )

    return answer


# ============================================================
# UPLOAD FILE TO GEMINI
# ============================================================

def upload_to_gemini(file_path):

    path = Path(file_path)

    if not path.exists():

        print()
        print("ERROR: File does not exist.")
        print(path)

        return None

    if not path.is_file():

        print()
        print("ERROR: This is not a file.")

        return None

    print()
    print(
        f"Uploading: {path.name}"
    )

    try:

        uploaded_file = client.files.upload(
            file=str(path)
        )

        print(
            "Upload successful."
        )

        return uploaded_file

    except Exception as error:

        print()
        print(
            "UPLOAD ERROR:"
        )

        print(error)

        return None


# ============================================================
# ANALYZE IMAGE
# ============================================================

def analyze_image():

    print()
    print("=" * 70)
    print("IMAGE ANALYSIS")
    print("=" * 70)

    file_path = input(
        "Enter image path: "
    ).strip()

    question = input(
        "What should Gemini do with this image? "
    ).strip()

    uploaded_file = upload_to_gemini(
        file_path
    )

    if uploaded_file is None:

        return

    try:

        response = client.models.generate_content(
            model=TEXT_MODEL,
            contents=[
                uploaded_file,
                question
            ]
        )

        answer = response.text

        display_answer(answer)

        save_text(
            answer,
            "image_analysis.txt"
        )

    except Exception as error:

        print()
        print(
            "IMAGE ANALYSIS ERROR:"
        )

        print(error)


# ============================================================
# ANALYZE PDF
# ============================================================

def analyze_pdf():

    print()
    print("=" * 70)
    print("PDF ANALYSIS")
    print("=" * 70)

    file_path = input(
        "Enter PDF path: "
    ).strip()

    question = input(
        "What should Gemini do with this PDF? "
    ).strip()

    uploaded_file = upload_to_gemini(
        file_path
    )

    if uploaded_file is None:

        return

    try:

        response = client.models.generate_content(
            model=TEXT_MODEL,
            contents=[
                uploaded_file,
                question
            ]
        )

        answer = response.text

        display_answer(answer)

        save_text(
            answer,
            "pdf_analysis.txt"
        )

    except Exception as error:

        print()
        print(
            "PDF ANALYSIS ERROR:"
        )

        print(error)


# ============================================================
# ANALYZE ANY FILE
# ============================================================

def analyze_file():

    print()
    print("=" * 70)
    print("FILE ANALYSIS")
    print("=" * 70)

    file_path = input(
        "Enter file path: "
    ).strip()

    question = input(
        "What should Gemini do with this file? "
    ).strip()

    uploaded_file = upload_to_gemini(
        file_path
    )

    if uploaded_file is None:

        return

    try:

        response = client.models.generate_content(
            model=TEXT_MODEL,
            contents=[
                uploaded_file,
                question
            ]
        )

        answer = response.text

        display_answer(answer)

        save_text(
            answer,
            "file_analysis.txt"
        )

    except Exception as error:

        print()
        print(
            "FILE ANALYSIS ERROR:"
        )

        print(error)


# ============================================================
# MULTIPLE FILES
# ============================================================

def analyze_multiple_files():

    print()
    print("=" * 70)
    print("MULTIPLE FILE ANALYSIS")
    print("=" * 70)

    print()
    print(
        "Enter file paths separated by commas."
    )

    files_input = input(
        "Files: "
    ).strip()

    question = input(
        "What should Gemini do with these files? "
    ).strip()

    file_paths = [
        item.strip()
        for item in files_input.split(",")
        if item.strip()
    ]

    uploaded_files = []

    for file_path in file_paths:

        uploaded_file = upload_to_gemini(
            file_path
        )

        if uploaded_file:

            uploaded_files.append(
                uploaded_file
            )

    if not uploaded_files:

        print(
            "No files were uploaded."
        )

        return

    try:

        contents = (
            uploaded_files
            + [question]
        )

        response = client.models.generate_content(
            model=TEXT_MODEL,
            contents=contents
        )

        answer = response.text

        display_answer(answer)

        save_text(
            answer,
            "multiple_files_analysis.txt"
        )

    except Exception as error:

        print()
        print(
            "MULTIPLE FILE ERROR:"
        )

        print(error)


# ============================================================
# IMAGE GENERATION
# ============================================================

def generate_image():

    print()
    print("=" * 70)
    print("AI IMAGE GENERATION")
    print("=" * 70)

    prompt = input(
        "Describe the image you want: "
    ).strip()

    if not prompt:

        print(
            "Please enter an image description."
        )

        return

    print()
    print(
        "Generating image..."
    )

    try:

        response = client.models.generate_content(
            model=IMAGE_MODEL,
            contents=prompt
        )

        image_saved = False

        for candidate in response.candidates:

            if not candidate.content:

                continue

            for part in candidate.content.parts:

                # --------------------------------------------
                # GENERATED IMAGE
                # --------------------------------------------

                if getattr(
                    part,
                    "inline_data",
                    None
                ):

                    image = part.as_image()

                    timestamp = datetime.now().strftime(
                        "%Y%m%d_%H%M%S"
                    )

                    output_path = (
                        OUTPUT_FOLDER
                        / f"generated_image_{timestamp}.png"
                    )

                    image.save(
                        output_path
                    )

                    print()
                    print("=" * 70)
                    print("IMAGE CREATED SUCCESSFULLY")
                    print("=" * 70)
                    print()
                    print(
                        f"Saved to:\n{output_path}"
                    )

                    image_saved = True

                # --------------------------------------------
                # TEXT
                # --------------------------------------------

                if getattr(
                    part,
                    "text",
                    None
                ):

                    print()
                    print(
                        "Gemini:"
                    )

                    print(
                        part.text
                    )

        if not image_saved:

            print()
            print(
                "Gemini did not return an image."
            )

    except Exception as error:

        print()
        print(
            "IMAGE GENERATION ERROR:"
        )

        print(error)


# ============================================================
# SAVE TEXT FILE
# ============================================================

def save_text(
    text,
    filename
):

    output_path = (
        OUTPUT_FOLDER
        / filename
    )

    try:

        output_path.write_text(
            text,
            encoding="utf-8"
        )

        print()
        print(
            f"Saved text file:\n{output_path}"
        )

    except Exception as error:

        print(
            "Could not save text:"
        )

        print(error)


# ============================================================
# CREATE PDF REPORT
# ============================================================

def create_pdf():

    print()
    print("=" * 70)
    print("CREATE PDF")
    print("=" * 70)

    title = input(
        "PDF title: "
    ).strip()

    print()
    print(
        "Enter the content for the PDF."
    )

    print(
        "Type END on a new line when finished."
    )

    print()

    lines = []

    while True:

        line = input()

        if line.strip().upper() == "END":

            break

        lines.append(
            line
        )

    content = "\n".join(
        lines
    )

    if not content.strip():

        print(
            "No content entered."
        )

        return

    timestamp = datetime.now().strftime(
        "%Y%m%d_%H%M%S"
    )

    pdf_path = (
        OUTPUT_FOLDER
        / f"gemini_report_{timestamp}.pdf"
    )

    try:

        document = SimpleDocTemplate(
            str(pdf_path),
            pagesize=A4,
            rightMargin=50,
            leftMargin=50,
            topMargin=50,
            bottomMargin=50
        )

        styles = getSampleStyleSheet()

        title_style = styles["Title"]

        title_style.alignment = TA_CENTER

        body_style = styles["BodyText"]

        body_style.leading = 16

        story = []

        # ----------------------------------------------------
        # TITLE
        # ----------------------------------------------------

        story.append(
            Paragraph(
                title,
                title_style
            )
        )

        story.append(
            Spacer(
                1,
                20
            )
        )

        # ----------------------------------------------------
        # DATE
        # ----------------------------------------------------

        story.append(
            Paragraph(
                "Generated: "
                + datetime.now().strftime(
                    "%Y-%m-%d %H:%M:%S"
                ),
                styles["Normal"]
            )
        )

        story.append(
            Spacer(
                1,
                20
            )
        )

        # ----------------------------------------------------
        # CONTENT
        # ----------------------------------------------------

        paragraphs = content.split(
            "\n"
        )

        for paragraph in paragraphs:

            if paragraph.strip():

                # Escape basic HTML characters
                safe_text = (
                    paragraph
                    .replace(
                        "&",
                        "&amp;"
                    )
                    .replace(
                        "<",
                        "&lt;"
                    )
                    .replace(
                        ">",
                        "&gt;"
                    )
                )

                story.append(
                    Paragraph(
                        safe_text,
                        body_style
                    )
                )

                story.append(
                    Spacer(
                        1,
                        8
                    )
                )

        document.build(
            story
        )

        print()
        print("=" * 70)
        print("PDF CREATED SUCCESSFULLY")
        print("=" * 70)
        print()
        print(
            f"Saved to:\n{pdf_path}"
        )

    except Exception as error:

        print()
        print(
            "PDF CREATION ERROR:"
        )

        print(error)


# ============================================================
# CREATE PDF FROM GEMINI RESPONSE
# ============================================================

def ask_gemini_and_create_pdf():

    print()
    print("=" * 70)
    print("GEMINI → PDF")
    print("=" * 70)

    question = input(
        "What should Gemini write about? "
    ).strip()

    if not question:

        return

    print()
    print(
        "Asking Gemini..."
    )

    try:

        response = client.models.generate_content(
            model=TEXT_MODEL,
            contents=question
        )

        answer = response.text

        print()
        print("=" * 70)
        print("GEMINI RESPONSE")
        print("=" * 70)
        print(answer)
        print("=" * 70)

        title = input(
            "Enter PDF title: "
        ).strip()

        if not title:

            title = "Gemini AI Report"

        timestamp = datetime.now().strftime(
            "%Y%m%d_%H%M%S"
        )

        pdf_path = (
            OUTPUT_FOLDER
            / f"gemini_report_{timestamp}.pdf"
        )

        document = SimpleDocTemplate(
            str(pdf_path),
            pagesize=A4,
            rightMargin=50,
            leftMargin=50,
            topMargin=50,
            bottomMargin=50
        )

        styles = getSampleStyleSheet()

        title_style = styles["Title"]

        title_style.alignment = TA_CENTER

        body_style = styles["BodyText"]

        body_style.leading = 16

        story = []

        story.append(
            Paragraph(
                title,
                title_style
            )
        )

        story.append(
            Spacer(
                1,
                20
            )
        )

        story.append(
            Paragraph(
                "Generated by Gemini",
                styles["Normal"]
            )
        )

        story.append(
            Spacer(
                1,
                20
            )
        )

        for paragraph in answer.split(
            "\n"
        ):

            if paragraph.strip():

                safe_text = (
                    paragraph
                    .replace(
                        "&",
                        "&amp;"
                    )
                    .replace(
                        "<",
                        "&lt;"
                    )
                    .replace(
                        ">",
                        "&gt;"
                    )
                )

                story.append(
                    Paragraph(
                        safe_text,
                        body_style
                    )
                )

                story.append(
                    Spacer(
                        1,
                        8
                    )
                )

        document.build(
            story
        )

        print()
        print("=" * 70)
        print("PDF CREATED")
        print("=" * 70)
        print()
        print(
            f"Saved to:\n{pdf_path}"
        )

    except Exception as error:

        print()
        print(
            "GEMINI PDF ERROR:"
        )

        print(error)


# ============================================================
# DISPLAY ANSWER
# ============================================================

def display_answer(answer):

    print()
    print("=" * 70)
    print("🤖 GEMINI")
    print("=" * 70)
    print()
    print(answer)
    print()
    print("=" * 70)


# ============================================================
# CLEAR CONVERSATION
# ============================================================

def clear_conversation():

    global conversation_history

    conversation_history = []

    print()
    print(
        "Conversation cleared."
    )


# ============================================================
# MAIN MENU
# ============================================================

def show_menu():

    print()
    print()
    print("=" * 70)
    print("                    🤖 GEMINI AI")
    print("=" * 70)
    print()
    print("  1. 💬 Chat with Gemini")
    print("  2. 🖼️  Analyze an image")
    print("  3. 📄 Analyze a PDF")
    print("  4. 📁 Analyze a file")
    print("  5. 📚 Analyze multiple files")
    print("  6. 🎨 Generate an image")
    print("  7. 📝 Ask Gemini and create PDF")
    print("  8. 📄 Create PDF manually")
    print("  9. 🗑️  Clear conversation")
    print("  0. ❌ Exit")
    print()
    print("=" * 70)


# ============================================================
# MAIN PROGRAM
# ============================================================

def main():

    print()
    print("=" * 70)
    print("           WELCOME TO YOUR GEMINI AI CHATBOT")
    print("=" * 70)
    print()
    print(
        "Your API key is configured."
    )
    print()

    while True:

        show_menu()

        choice = input(
            "Choose an option: "
        ).strip()

        # ----------------------------------------------------
        # CHAT
        # ----------------------------------------------------

        if choice == "1":

            print()
            print(
                "Type 'exit' to return to the main menu."
            )

            while True:

                question = input(
                    "\nYou: "
                ).strip()

                if question.lower() == "exit":

                    break

                if not question:

                    continue

                try:

                    answer = chat_with_gemini(
                        question
                    )

                    display_answer(
                        answer
                    )

                except Exception as error:

                    print()
                    print(
                        "CHAT ERROR:"
                    )

                    print(error)

        # ----------------------------------------------------
        # IMAGE ANALYSIS
        # ----------------------------------------------------

        elif choice == "2":

            analyze_image()

        # ----------------------------------------------------
        # PDF ANALYSIS
        # ----------------------------------------------------

        elif choice == "3":

            analyze_pdf()

        # ----------------------------------------------------
        # FILE ANALYSIS
        # ----------------------------------------------------

        elif choice == "4":

            analyze_file()

        # ----------------------------------------------------
        # MULTIPLE FILES
        # ----------------------------------------------------

        elif choice == "5":

            analyze_multiple_files()

        # ----------------------------------------------------
        # IMAGE GENERATION
        # ----------------------------------------------------

        elif choice == "6":

            generate_image()

        # ----------------------------------------------------
        # GEMINI → PDF
        # ----------------------------------------------------

        elif choice == "7":

            ask_gemini_and_create_pdf()

        # ----------------------------------------------------
        # MANUAL PDF
        # ----------------------------------------------------

        elif choice == "8":

            create_pdf()

        # ----------------------------------------------------
        # CLEAR CHAT
        # ----------------------------------------------------

        elif choice == "9":

            clear_conversation()

        # ----------------------------------------------------
        # EXIT
        # ----------------------------------------------------

        elif choice == "0":

            print()
            print(
                "Thank you for using Gemini AI. 👋"
            )
            print()

            break

        # ----------------------------------------------------
        # INVALID
        # ----------------------------------------------------

        else:

            print()
            print(
                "Invalid option. Please choose 0-9."
            )


# ============================================================
# START
# ============================================================

if __name__ == "__main__":

    main()
```
